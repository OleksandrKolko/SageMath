__Знайдіть к-ть різних розфарбувань кубика Рубика 2х2х2, використовуючи 10 кольорів.__


![rubik cube](../docs/labs_assets/rubik_cube.jpg)

Два розфарбування називатимуться однаковими, якщо існує послідовність рухів кубика рубика, яка одне розфарбування переводить в інше. 

Не обовʼязково використовувати всі 10 кольорів, тобто допустимо пофарбувати весь кубик в один колір. 

___Приклад___: існує 183 різних розфарбувань, використовуючи 2 кольори. 

```python
# Імпортуємо потрібні бібліотеки
G = SymmetricGroup(24)

# Задаємо генератори групи симетрій кубика
g1 = G('(1, 2, 4, 3)(5, 6, 8, 7)(9, 10, 12, 11)(13, 14, 16, 15)(17, 18, 20, 19)(21, 22, 24, 23)')
g2 = G('(1, 5, 9, 13, 17, 21)(2, 6, 10, 14, 18, 22)(3, 7, 11, 15, 19, 23)(4, 8, 12, 16, 20, 24)')
g3 = G('(1, 3)(2, 4)(5, 7)(6, 8)(9, 11)(10, 12)(13, 15)(14, 16)(17, 19)(18, 20)(21, 23)(22, 24)')
g4 = G('(1, 9)(2, 10)(3, 11)(4, 12)(5, 13)(6, 14)(7, 15)(8, 16)(17, 21)(18, 22)(19, 23)(20, 24)')
g5 = G('(5, 9)(6, 10)(7, 11)(8, 12)(13, 17)(14, 18)(15, 19)(16, 20)(21, 23)(22, 24)')
g6 = G('(13, 21)(14, 22)(15, 23)(16, 24)')

# Створюємо групу симетрій кубика
H = G.subgroup([g1, g2, g3, g4, g5, g6])

# Підраховуємо розмір групи симетрій
print("Розмір групи симетрій:", H.order())

# Кількість кольорів
num_colors = 10

# Використовуємо теорему Бьорнсайда
def fixed_colorings(perm, num_colors):
    """Обчислює кількість фіксованих розфарбувань для даної перестановки."""
    cycles = perm.cycle_type()  # Тип циклів у перестановці
    return num_colors**sum(cycles)

# Підрахунок кількості унікальних розфарбувань
unique_colorings = sum(fixed_colorings(h, num_colors) for h in H) / H.order()

print("Кількість унікальних розфарбувань:", unique_colorings)
```
