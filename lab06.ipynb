## Завдання 1

В групі рухів правильного восьмикутника задайте циклічну підгрупу порядку 4. Побудуйте для неї ліві і праві класи суміжності та перевірте чи буде ця група нормальною. 

```python
import time

def left_classes(group, subgroup):
    l_classes = []
    for x in group:
        new_class = []
        for y in subgroup:
            new_class.append(x * y)
        if new_class not in l_classes:
            new_class = sorted(new_class)
            l_classes.append(new_class)
    return l_classes

def right_classes(group, subgroup):
    r_classes = []
    for x in group:
        new_class = []
        for y in subgroup:
            new_class.append(y * x)
        if new_class not in r_classes:
            new_class = sorted(new_class)
            r_classes.append(new_class)
    return r_classes
    
D = DihedralGroup(8)
elems = []
for el in D:
    if el.order() == 4:
        elems.append(el)
G = D.subgroup(elems)
print(G)

l_classes = left_classes(D, G)

r_classes = right_classes(D, G)

print("Left classes:")
for el in l_classes:
    print(el)
print("Right classes:")
for el in r_classes:
    print(el)

print()
if l_classes == r_classes:
    print("Left and right classes are equal, so this subgroup is normal.")
print(f'Is subgroup normal: {G.is_normal()}')
```
