## Завдання 2 

Розгляньте групу матриць за множенням: 

$$G = \left\langle \left(\begin{array}{rr}
1 & -1 \\
-1 & 2
\end{array}\right), \left(\begin{array}{rr}
3 & 2 \\
2 & 1
\end{array}\right) \right\rangle$$

Дайте відповідь на наступні питання: 

1. _Чи є група G скінченною?_
2. _Чи є група G комутативною?_
3. _Який вигляд мають елементи групи G?_
4. _Які порядки мають елементи групи G?_ 
5. _Чому ізоморфна група G?_


__Відповіді__: 
1. _Ні_
2. _Так_
3. _Загальний вигляд є дуже громізким, але кожен елемент схожий на матричне зображення чисел Фібоначчі_
4. _Елементи маю нескінченні порядки_
5. _Точно не знаю, але може якійсь підмонжині з групи цілих чисел, які відповідають числам Фібоначчі_

```python
var("k m")
A = Matrix([[1, -1],
           [-1, 2],])

B = Matrix([[3, 2],
           [2, 1]])

print("Вигляд порядків матриці A:")
for i in range(1, 5):
    print(A^i)
    print()

print("Вигляд порядків матриці B:")
for i in range(1, 5):
    print(B^i)
    print()

print("Вигляд порядків матриці (A * B)^n:")
for i in range(1, 5):
    print((A * B)^i)
    print()

print("Загальний вигляд для елементів A^k * B^m")
print(A^k * B^m)

if (A^k * B^m) == (B^m * A^k):
    print("Група є комутативною")
```
